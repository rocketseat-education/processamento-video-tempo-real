```python
# Tracking de Objetos em Vídeo
import cv2
import time
```

```python
# Criar Tracker

def criar_tracker(tipo):
    if tipo == "MOSSE":
        return cv2.legacy.TrackerMOSSE_create()
    elif tipo == "CSRT":
        return cv2.TrackerCSRT_create()
    elif tipo == "KCF":
        return cv2.TrackerKCF_create()
    elif tipo == "MIL":
        return cv2.TrackerMIL_create()
    elif tipo == "BOOSTING":
        return cv2.legacy.TrackerBoosting_create()
```

```python
# Escolha do Tracker
print("Escolha o tracker")
print("1 - MOSSE       | Mais Rápido")
print("2 - CSRT        | Mais Preciso")
print("3 - KCF         | Meio Termo (Rápido e Preciso)")
print("4 - MIL         | Meio Termo (Rápido e Preciso)")
print("5 - BOOST       | Mais Antigo")

opcao = input("Digite o número do tracker")

if opcao == "1":
    tipo_tracker = "MOSSE"
elif opcao == "2":
    tipo_tracker = "CSRT"    
elif opcao == "3":
    tipo_tracker = "KCF"   
elif opcao == "4":
    tipo_tracker = "MIL" 
elif opcao == "5":
    tipo_tracker = "BOOSTING" 

tracker = criar_tracker(tipo_tracker)
```

```python
# Abertura da cam
camera = cv2.VideoCapture(0)

if not camera.isOpened():
    print("Erro ao abrir webcam.")
    exit()
```

```python
# Loop Pré-Visualização

print("A webcam será exibida")
print("Pressione S para selecionar o objeto")
print("Pressione ESC para sair")

bbox = None

while True:
    ret, frame = camera.read()

    if not ret:
        print("Erro ao capturar frame.")
        break
    
    cv2.putText(
        frame,
        "Pressione S para selecionar o objeto",
        (20, 75),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 255, 0),
        2
    )
    cv2.putText(
        frame,
        f"Tracker: {tipo_tracker}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 0),
        2
    )

    cv2.imshow("Preview da Webcam", frame)

    tecla = cv2.waitKey(1) & 0xFF

    if tecla == 27:
        camera.release()
        cv2.destroyAllWindows()
        exit()
    elif tecla == ord("s") or tecla == ord("S"):
        bbox = cv2.selectROI(
            "Selecione o Objeto",
            frame,
            True
        )
        cv2.destroyWindow("Selecione o Objeto")
        cv2.destroyWindow("Preview da Webcam")
        break 
```
