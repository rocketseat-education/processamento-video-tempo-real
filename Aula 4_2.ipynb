```python
import cv2
import os
```

```python
pasta_videos = "Videos"
# nome_video = "Foguete (480p).mp4"
nome_video = "Starship (360p).mp4"
caminho_video = os.path.join(pasta_videos,nome_video)
```

```python
if not os.path.exists(caminho_video):
    print("Arquivo de Vídeo não encontrado.")
    exit()

video = cv2.VideoCapture(caminho_video)

if not video.isOpened():
    print("Erro ao abrir o vídeo.")
    exit()
```

```python
# Informações do vídeo

largura = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
altura = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(video.get(cv2.CAP_PROP_FPS))
total_frames = int(video.get(cv2.CAP_PROP_FRAME_COUNT))

duracao = total_frames/fps


# 24 fps  e 1 segundo tem 1000 milisegundos
# 1000 / 24 => 41,6 milisegundos 
tempo_espera = int(1000/fps)

print("Vídeo carregado com sucesso!")
print("Caminho:", caminho_video)
print("Largura:", largura)
print("Altura:", altura)
print("FPS:", fps)
print("Total de frames:", total_frames)
print("Duração aproximada:", duracao, "segundos")
```

```python
#loop Principal

while True:
    ret, frame = video.read()
    if ret == False:
        print("Fim do vídeo")
        break

    # Texto no vídeo
    texto_resolucao = f"Resolucao:{largura}x{altura}"
    texto_fps = f"FPS: {fps}"

    cv2.putText(frame, texto_resolucao, (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
    cv2.putText(frame, texto_fps, (20, 75), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)

    cv2.imshow("Aula 4.2", frame)

    tecla = cv2.waitKey(tempo_espera) & 0xFF 

    if tecla == 27:
        print("Programa Encerrado pelo Usuário")
        break

video.release()
cv2.destroyAllWindows()
```
