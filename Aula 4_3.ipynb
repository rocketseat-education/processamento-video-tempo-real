```python
import cv2
import os
```

```python
# Configurações Iniciais

indice_camera = 0

largura = 1280
altura = 720

fps = 30

nome_arquivo = "video_gravado.mp4"
caminho_saida = os.path.join(os.getcwd(),nome_arquivo)
```

```python
# Abrir webcam
camera = cv2.VideoCapture(indice_camera)
if not camera.isOpened():
    print("Erro ao abrir a webcam.")
    exit()
```

```python
# Configuração de Resolução
camera.set(cv2.CAP_PROP_FRAME_WIDTH,largura)
camera.set(cv2.CAP_PROP_FRAME_HEIGHT,altura)

largura_real = int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
altura_real = int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))

print("Webcam aberta com sucesso.")
print("Resolução real:", largura_real, "x", altura_real)
print("FPS configurado:", fps)
print("Arquivo será salvo em:")
print(caminho_saida)
```

```python
# Configurar o Codec

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

# "mp4v"	MP4 básico
# "XVID"	AVI
# "MJPG"	Motion JPEG
# "H264"	H.264
# "avc1"	AVC/H264
```

```python
# Gravação de Vídeo
gravador = cv2.VideoWriter(
    caminho_saida,
    fourcc,
    fps,
    (largura_real, altura_real)
)
if not gravador.isOpened():
    print("Erro ao criar o arquivo de vídeo.")
    camera.release
    exit()
```

```python
# Loop principal
while True:
    ret, frame = camera.read()
    if not ret:
        print("Erro ao capturar frame.")
        break
    texto_1 = "Aula 4.3 - Gravando Vídeo"
    texto_2 = "Pressione Esc para encerrar"
    texto_3 = f"Resolucao: {largura_real}x{altura_real}"
    texto_4 = f"FPS:{fps}"

    cv2.putText(frame, texto_1,(20,40),
                cv2.FONT_HERSHEY_SIMPLEX,0.8,
                (255,255,255),2)
    cv2.putText(frame, texto_2,(20,75),
                cv2.FONT_HERSHEY_SIMPLEX,0.8,
                (255,255,255),2)
    cv2.putText(frame, texto_3,(20,110),
                cv2.FONT_HERSHEY_SIMPLEX,0.8,
                (255,255,255),2)
    cv2.putText(frame, texto_4,(20,145),
                cv2.FONT_HERSHEY_SIMPLEX,0.8,
                (255,255,255),2)

    gravador.write(frame)

    cv2.imshow("Aula 4.3", frame)

    tecla = cv2.waitKey(1) & 0xFF

    if tecla == 27:
        print("Gravação encerrada")
        break

camera.release()
gravador.release()
cv2.destroyAllWindows()

print("Video Salvo com sucesso!")
```
