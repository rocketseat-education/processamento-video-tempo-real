```python
import cv2
```

```python
camera = cv2.VideoCapture(0)
```

```python
fps = camera.get(cv2.CAP_PROP_FPS)
print("FPS:", fps)
```

```python
largura = camera.get(cv2.CAP_PROP_FRAME_WIDTH)
altura = camera.get(cv2.CAP_PROP_FRAME_HEIGHT)

print("Largura:", largura)
print("Altura", altura)
```

```python
#Loop principal
while True:
    ret, frame = camera.read()
    if ret == False:
        print("Erro ao capturar vídeo")
        break
    cv2.imshow("Webcam OpenCV", frame)

    tecla = cv2.waitKey(1)

    if tecla == 27:
        break

camera.release()
cv2.destroyAllWindows()
```
